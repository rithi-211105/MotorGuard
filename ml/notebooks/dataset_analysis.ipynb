# Dataset Exploratory Data Analysis (EDA)
## CWRU Motor Bearing Vibration Feature Dataset (`feature_time_48k_2048_load_1.csv`)

This notebook provides a complete exploratory data analysis of the motor vibration feature dataset, evaluating features, class balance, data leakage risks, and suitability for anomaly detection & multi-class fault classification.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load dataset
filename = 'feature_time_48k_2048_load_1.csv'
df = pd.read_csv(filename)
print(f"Successfully loaded {filename} with shape {df.shape}")
```

### 1. Dataset Overview & Data Types

```python
print("=== DATASET INFO ===")
print(df.info())
print("\n=== FIRST 5 ROWS ===")
df.head()
```

### 2. Missing Values & Duplicate Check

```python
missing = df.isnull().sum()
duplicates = df.duplicated().sum()
print(f"Missing Values per Column:\n{missing}")
print(f"\nDuplicate Rows Count: {duplicates}")
```

### 3. Target Column & Class Distribution

```python
class_counts = df['fault'].value_counts()
print("=== CLASS DISTRIBUTION ===")
print(class_counts)

plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='fault', palette='Blues_r', order=class_counts.index)
plt.title('Class Sample Counts (Load 1)')
plt.xticks(rotation=45, ha='right')
plt.show()
```

### 4. Descriptive Statistics

```python
df.describe().T[['mean', 'std', 'min', '50%', 'max']]
```

### 5. Feature Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(9, 7))
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='Blues', vmin=-1, vmax=1)
plt.title('Feature Correlation Matrix')
plt.show()
```

### 6. Feature Distributions by Fault Class

```python
features = [c for c in df.columns if c != 'fault']
plt.figure(figsize=(14, 10))
for idx, feat in enumerate(features, 1):
    plt.subplot(3, 3, idx)
    sns.boxplot(data=df, x='fault', y=feat, palette='Blues', order=class_counts.index)
    plt.xticks(rotation=60, ha='right', fontsize=8)
plt.tight_layout()
plt.show()
```
